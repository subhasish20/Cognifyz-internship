### Analyze the distribution of the target variable("Aggregate rating") and identify any class imbalances.



```python
df.dtypes
```

```python
# Plot the distribution of the target variable "Aggregate rating"
plt.figure(figsize=(10,6))
sns.histplot(df['Aggregate rating'], kde=True)
plt.title('Distribution of Aggregate Rating')
plt.xlabel('Aggregate Rating')
plt.ylabel('Frequency')
plt.grid()
plt.show()
```

```python
# Checking for skewness
skewness = df['Aggregate rating'].skew()
print(f"Skewness of Aggregate Rating: {skewness}")
```

```python
# Optional: Convert Aggregate Rating to categorical bins
bins = [0, 2, 3, 4, 5]
labels = ['1-2 stars', '2-3 stars', '3-4 stars', '4-5 stars']
df['Rating Category'] = pd.cut(df['Aggregate rating'], bins=bins, labels=labels)

# Plot the counts of each category
plt.figure(figsize=(8,6))
sns.countplot(data=df, x='Rating Category')
plt.title('Class Distribution of Rating Categories')
plt.xlabel('Rating Categories')
plt.ylabel('Count')
plt.show()
```
